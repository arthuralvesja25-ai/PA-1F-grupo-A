```python
#entrada
numero1 = int(input("Digite o primeiro número: "))
numero2 = int(input("Digite o segundo número: "))

#saida

if numero1 > numero2:
    print(f"O primeiro número ({numero1}) é maior que o segundo número ({numero2}).")
elif numero2 > numero1:
    print(f"O segundo número ({numero2}) é maior que o primeiro número ({numero1}).")
else:
    print(f"Os dois números são iguais ({numero1}).")
```

```python
# entrada
media = float(input("Digite a média final do aluno: "))

#saida

if media >= 7.0:
    print("Aprovado")
elif 5.0 <= media < 7.0:
    print("Recuperação")
else:
    print("Reprovado")
```

```python
#entrada
numero_dia = int(input("Digite um número de 1 a 7: "))

#processamento

dias_semana = {
    1: "Domingo",
    2: "Segunda-feira",
    3: "Terça-feira",
    4: "Quarta-feira",
    5: "Quinta-feira",
    6: "Sexta-feira",
    7: "Sábado"
}

#saida

if 1 <= numero_dia <= 7:
    print(f"O dia correspondente é: {dias_semana[numero_dia]}")
else:
    print("Número inválido. Por favor, digite um número entre 1 e 7.")
```

```python
 #entrada
ano = int(input("Digite um ano: "))

# processamento e saída
if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
    print(f"O ano {ano} é bissexto.")
else:
    print(f"O ano {ano} NÃO é bissexto.")
```

```python
# Entrada
valor_compra = float(input("Digite o valor da compra: "))

# Processamento
desconto = 0
if valor_compra >= 100:
    desconto = 0.10  # 10% de desconto
elif valor_compra >= 50:
    desconto = 0.05  # 5% de desconto
# else: desconto permanece 0 (sem desconto)

valor_com_desconto = valor_compra * (1 - desconto)

# Saída
print(f"Valor original da compra: R$ {valor_compra:.2f}")
print(f"Desconto aplicado: {desconto * 100:.0f}%")
print(f"Valor final com desconto: R$ {valor_com_desconto:.2f}")
```
