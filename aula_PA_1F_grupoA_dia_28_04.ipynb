```python

#escrever um programa que o usuario digita uma palavra
# e o programa verifica se é palindrome


#entrada
palavra = str (input("digite uma palavra:"))
invertida = ""

for i in (palavra):

  invertida =  i + invertida

  if invertida == palavra:

    print(f"(palavra) é uma palindrome")
  else:
    print(f"(palavra) não é uma palindrome")
```
