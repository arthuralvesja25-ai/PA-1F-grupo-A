```python

import random

numero_secreto = random.randint(0, 10)

tentativas = 0

print("=== JOGO DE ADIVINHAÇÃO ===")
print("Adivinhe o número de 0 a 10")
print("Você tem 3 tentativas")

while tentativas < 3:

    palpite = int(input("Digite seu palpite: "))
    tentativas += 1

    if palpite == numero_secreto:
        print("Parabéns! Você acertou!")
        break

    elif palpite > numero_secreto:
        print("Seu palpite é MAIOR que o número secreto.")

        diferenca = abs(numero_secreto - palpite)

        if diferenca == 1:
            print("Quente")
        elif diferenca == 2:
            print("Morno")
        elif diferenca == 3:
            print("Frio")
        else:
            print("Muito frio")

    elif palpite < numero_secreto:
        print("Seu palpite é MENOR que o número secreto.")

        diferenca = abs(numero_secreto - palpite)

        if diferenca == 1:
            print("Quente")
        elif diferenca == 2:
            print("Morno")
        elif diferenca == 3:
            print("Frio")
        else:
            print("Muito frio")

if palpite != numero_secreto:
    print("Você perdeu!")
    print("O número secreto era:", numero_secreto)
```
