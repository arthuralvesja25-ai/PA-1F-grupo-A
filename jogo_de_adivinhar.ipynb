```python
import random

# Gera um número secreto entre 0 e 10
numero_secreto = random.randint(0, 10)

max_tentativas = 3
tentativas = 0

print("Bem-vindo ao jogo de adivinhação!")
print("Estou pensando em um número entre 0 e 10.")
print(f"Você tem {max_tentativas} tentativas para adivinhar.")

while tentativas < max_tentativas:
    try:
        palpite = int(input("Digite seu palpite (0 a 10): "))

        if palpite < 0 or palpite > 10:
            print("Digite um número entre 0 e 10.")
            continue

        tentativas += 1

        if palpite == numero_secreto:
            print(f"Parabéns! Você acertou o número {numero_secreto}!")
            break

        # Maior ou menor
        if palpite > numero_secreto:
            print("Seu palpite é MAIOR que o número secreto.")
        else:
            print("Seu palpite é MENOR que o número secreto.")

        # Quente, morno ou frio
        diferenca = abs(numero_secreto - palpite)

        if diferenca == 1:
            print("Quente!")
        elif diferenca == 2:
            print("Morno!")
        elif diferenca == 3:
            print("Frio!")
        else:
            print("Muito frio!")

        print(f"Restam {max_tentativas - tentativas} tentativas.")

    except ValueError:
        print("Digite apenas números inteiros!")

if tentativas == max_tentativas and palpite != numero_secreto:
    print(f"Você perdeu! O número secreto era {numero_secreto}.")

print("Fim do jogo!")
```
