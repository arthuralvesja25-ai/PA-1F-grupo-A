```python
def executar_sistema_votacao():
    # 1. Estrutura com os novos candidatos solicitados e a opção de Voto em Branco
    candidatos = {
        1: {"nome": "Alpha", "votos": 0},
        2: {"nome": "Beta", "votos": 0},
        3: {"nome": "Sigma", "votos": 0},
        4: {"nome": "Chade", "votos": 0},
        0: {"nome": "Voto em Branco", "votos": 0}
    }

    # Contadores gerais do sistema
    total_eleitores_cadastrados = 0
    total_eleitores_votaram = 0
    votos_opcionais = 0
    votos_obrigatorios = 0

    print("=== SISTEMA DE VOTAÇÃO ELETRÔNICA ===")

    while True:
        print("\n--- Cadastro de Eleitor ---")
        nome = input("Digite o nome do eleitor: ").strip()

        # Validação: Impede idades inválidas
        while True:
            try:
                idade = int(input(f"Digite a idade de {nome}: "))
                if 0 <= idade <= 120:
                    break
                print("Idade inválida! Por favor, digite uma idade entre 0 e 120 anos.")
            except ValueError:
                print("Entrada inválida! Digite apenas números inteiros para a idade.")

        total_eleitores_cadastrados += 1

        # 2. Verificação da situação eleitoral por idade
        if idade < 16:
            print(f"Situação: {nome} tem {idade} anos e NÃO PODE VOTAR.")
            apto_a_votar = False
            tipo_voto = None
        elif 16 <= idade <= 17:
            print(f"Situação: {nome} tem {idade} anos. Voto OPCIONAL.")
            apto_a_votar = True
            tipo_voto = "opcional"
        else:
            print(f"Situação: {nome} tem {idade} anos. Voto OBRIGATÓRIO.")
            apto_a_votar = True
            tipo_voto = "obrigatorio"

        # 3. Processo de votação para eleitores aptos
        if apto_a_votar:
            quer_votar = "s"
            if tipo_voto == "opcional":
                quer_votar = input("Deseja exercer seu direito de voto? (S/N): ").strip().lower()

            if quer_votar == "s":
                total_eleitores_votaram += 1
                if tipo_voto == "opcional":
                    votos_opcionais += 1
                else:
                    votos_obrigatorios += 1

                print("\n--- CANDIDATOS DISPONÍVEIS ---")
                for numero, info in candidatos.items():
                    print(f"[{numero}] - {info['nome']}")

                # Validação: Permite nova tentativa se o candidato for inválido
                while True:
                    try:
                        voto = int(input("Digite o número do seu candidato: "))
                        if voto in candidatos:
                            candidatos[voto]["votos"] += 1
                            print(f"Voto confirmado para: {candidatos[voto]['nome']}!")
                            break
                        else:
                            print("Número inválido! Esse candidato não existe.")
                    except ValueError:
                        print("Entrada inválida! Digite apenas o número do candidato.")
            else:
                print(f"{nome} optou por não votar.")

        # Pergunta se deseja continuar o cadastro de novos eleitores
        continuar = input("\nDeseja cadastrar outro eleitor? (S/N): ").strip().lower()
        if continuar != "s":
            break

    # --- RELATÓRIO FINAL ---
    print("\n" + "="*40)
    print("           RESULTADO DA VOTAÇÃO           ")
    print("="*40)
    print(f"Total de eleitores cadastrados: {total_eleitores_cadastrados}")
    print(f"Total de eleitores que votaram: {total_eleitores_votaram}")
    print(f"Quantidade de votos opcionais praticados: {votos_opcionais}")
    print(f"Quantidade de votos obrigatórios praticados: {votos_obrigatorios}")
    print("-"*40)

    # Exibe votos e o percentual correspondente
    print("Votos por Candidato:")
    for numero, info in candidatos.items():
        percentual = (info["votos"] / total_eleitores_votaram * 100) if total_eleitores_votaram > 0 else 0
        print(f" - {info['nome']} (Nº {numero}): {info['votos']} voto(s) ({percentual:.2f}%)")

    print("-"*40)

    # 4. Identificação do vencedor ou empate (ignora o voto em branco para a vitória)
    candidatos_validos = {k: v for k, v in candidatos.items() if k != 0}

    if total_eleitores_votaram == 0:
        print("Nenhum voto foi registrado na urna.")
    else:
        maior_votacao = max(candidatos_validos[c]['votos'] for c in candidatos_validos)
        vencedores = [info["nome"] for k, info in candidatos_validos.items() if info["votos"] == maior_votacao]

        if maior_votacao == 0:
            print("Não houve votos nos candidatos principais (apenas votos em branco).")
        elif len(vencedores) > 1:
            print(f"Houve um EMPATE entre: {', '.join(vencedores)} com {maior_votacao} voto(s) cada!")
        else:
            print(f"Candidato VENCEDOR: {vencedores[0]} com {maior_votacao} voto(s)!")
    print("="*40)

# Executa o sistema
executar_sistema_votacao()
```
