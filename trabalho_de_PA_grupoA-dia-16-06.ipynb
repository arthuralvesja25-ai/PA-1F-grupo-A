```python

num = int(input("Digite um número inteiro positivo para a Soma dos Números: "))

if num > 0:
    soma = 0

    for i in range(1, num + 1):
        soma += i

    print(f"Resultado: {soma}")

elif num == 0:
    print("O número deve ser maior que zero.")

else:
    print("Por favor, digite um número inteiro positivo.")
```

```python

num_tabuada = int(input("Digite um número inteiro para a Tabuada: "))

if num_tabuada > 0:
    print(f"\nTabuada de {num_tabuada}:")
    for i in range(1, 11):
        print(f"{num_tabuada} x {i} = {num_tabuada * i}")

elif num_tabuada == 0:
    print("Tabuada de 0:")
    for i in range(1, 11):
        print(f"0 x {i} = 0")

else:
    print(f"\nTabuada de {num_tabuada}:")
    for i in range(1, 11):
        print(f"{num_tabuada} x {i} = {num_tabuada * i}")


```

```python

num_fatorial = int(input("Digite um número inteiro positivo para o Fatorial: "))

if num_fatorial < 0:
    print("Por favor, digite um número inteiro positivo.")

elif num_fatorial == 0:
    print("Fatorial = 1")

else:
    fatorial = 1
    for i in range(1, num_fatorial + 1):
        fatorial *= i

    print(f"Fatorial = {fatorial}")
```
