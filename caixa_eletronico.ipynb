```python
saldo = 1000.00

print("Bem-vindo ao Caixa Eletrônico!")

while True:
  print("\n--- Menu ---")
  print("1. Ver Saldo")
  print("2. Depositar")
  print("3. Sacar")
  print("4. Sair")

  opcao = input("Escolha uma opção: ")

  if opcao == '1':
    print(f"Seu saldo atual é: R${saldo:.2f}")
  elif opcao == '2':
    try:
      valor_deposito = float(input("Digite o valor a depositar: R$"))
      if valor_deposito > 0:
        saldo += valor_deposito
        print(f"Depósito de R${valor_deposito:.2f} realizado com sucesso. Novo saldo: R${saldo:.2f}")
      else:
        print("O valor do depósito deve ser positivo.")
    except ValueError:
      print("Valor inválido. Por favor, digite um número.")
  elif opcao == '3':
    try:
      valor_saque = float(input("Digite o valor a sacar: R$"))
      if valor_saque > 0:
        if saldo >= valor_saque:
          saldo -= valor_saque
          print(f"Saque de R${valor_saque:.2f} realizado com sucesso. Novo saldo: R${saldo:.2f}")
        else:
          print("Saldo insuficiente.")
      else:
        print("O valor do saque deve ser positivo.")
    except ValueError:
      print("Valor inválido. Por favor, digite um número.")
  elif opcao == '4':
    print("Obrigado por usar o Caixa Eletrônico. Volte sempre!")
    break
  else:
    print("Opção inválida. Por favor, escolha uma opção de 1 a 4.")
```
