```python
#entrada
idade = int(input("Digite a idade da pessoa: "))


#saida
if idade <= 17:
    print("Menor de idade")
elif 18 <= idade <= 59:
    print("Adulto")
else:
    print("Idoso")
```

```python
#entrada
valor_compra = float(input("Digite o valor da compra: R$"))

# processamento

if valor_compra <= 100:
    valor_final = valor_compra
elif 101 <= valor_compra <= 500:
    valor_final = valor_compra * 0.90  # 10% de desconto
else:
    valor_final = valor_compra * 0.80  # 20% de desconto


#saida
print(f"O valor final da compra com desconto é: R${valor_final:.2f}")
```

```python
# entrada
lado1 = float(input("Digite o valor do primeiro lado: "))
lado2 = float(input("Digite o valor do segundo lado: "))
lado3 = float(input("Digite o valor do terceiro lado: "))

#saida

if (lado1 + lado2 > lado3) and (lado1 + lado3 > lado2) and (lado2 + lado3 > lado1):
    print("Os valores podem formar um triângulo válido.")
else:
    print("Os valores NÃO podem formar um triângulo válido.")
```

```python
# entrada
media = float(input("Digite a média final do aluno: "))

#saida

if media >= 7.0:
    print("Aprovado")
elif 5.0 <= media < 7.0:
    print("Recuperação")
else:
    print("Reprovado")
```

```python
#entrada
pontuacao = int(input("Digite a pontuação do funcionário (0 a 10): "))
salario = float(input("Digite o salário do funcionário: R$"))

#processamento

bonus = 0.0

if 8 <= pontuacao <= 9:
    bonus = salario * 0.10  # 10% de bônus
elif pontuacao == 10:
    bonus = salario * 0.15  # 15% de bônus

salario_final = salario + bonus

#saida

print(f"O valor do bônus é: R${bonus:.2f}")
print(f"O salário final com bônus é: R${salario_final:.2f}")
```

```python
#entrada

energia = int(input("Digite a energia do personagem (0 a 100): "))

#processamento

if energia == 0:
    print("Morto")
elif 1 <= energia <= 30:
    print("Crítico")
elif 31 <= energia <= 70:
    print("Normal")
elif 71 <= energia <= 100:
    print("Saudável")
else:

    #saida
    print("Valor de energia inválido. Por favor, digite um número entre 0 e 100.")
```
