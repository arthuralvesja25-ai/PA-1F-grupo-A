escrever um programa que calcula a media das notas do aluno.

entrada: nome,idade,e as 04 notas do bimestre

processamento: calcular a media

saida: mostrar o nome do aluno e a media obtida

```python
# entrada

nome = input("nome do aluno: ")
idade = int(input("idade do aluno: "))
n1 = float(input("nota do primeiro bimestre: "))
n2 = float(input("nota do segundo bimestre: "))
n3 = float(input("nota do terceiro bimestre: "))
n4 = float(input("nota do quarto bimestre: "))

# processamento

media = (n1 + n2 + n3 + n4) / 4

# saida
print(f"o aluno {nome} de {idade} anos ficou com a media {media} de média")
```
